```python
marks = []
for i in range(5):
    mark = int(input(f"Enter marks for subject {i+1}: "))
    marks.append(mark)

total_marks = sum(marks)
percentage = total_marks / 5
print("Total Marks:", total_marks)
print("Percentage:", percentage)

match percentage:
    case p if p > 85:
        print("Grade A")
    case p if 85 >= p >= 75:
        print("Grade B")
    case p if 75 > p >= 50:
        print("Grade C")
    case p if 50 > p >= 30:
        print("Grade D")
    case p if p < 30:
        print("Reappear")
```
