```python
class LibraryBook:
    """
    Represents a library book with encapsulated attributes and methods for borrowing and returning.
    """

    def __init__(self, title, author, is_available=True):
        self.__title = title
        self.__author = author
        self.__is_available = is_available

    def get_title(self):
        """Returns the book's title."""
        return self.__title

    def get_author(self):
        """Returns the book's author."""
        return self.__author

    def is_available(self):
        """Checks if the book is available."""
        return self.__is_available

    def borrow(self):
        """Borrows the book if it's available, updates availability."""
        if self.is_available():
            self.__is_available = False
            print(f"{self.get_title()} by {self.get_author()} has been borrowed.")
        else:
            print(f"{self.get_title()} is not currently available.")

    def return_book(self):
        """Returns the book, updates availability."""
        self.__is_available = True
        print(f"{self.get_title()} by {self.get_author()} has been returned.")

# Create library book instances and test functionality
book1 = LibraryBook("The Hitchhiker's Guide to the Galaxy", "Douglas Adams")
book2 = LibraryBook("Pride and Prejudice", "Jane Austen", False)

book1.borrow()
book2.borrow()  # Book unavailable

book2.return_book()
book2.borrow()  # Now available
```
