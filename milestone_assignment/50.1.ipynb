```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
df = pd.read_csv('lung_cancer_dataset.csv')
```

```python
# Display the first few rows of the dataset
print(df.head())

# Check for missing values
print(df.isnull().sum())

# Summary statistics
print(df.describe())

# Check the data types
print(df.info())
```

```python
# Handling missing values (if any)
df.fillna(df.mean(), inplace=True)  # Example of filling missing values with mean
```

```python
# Age distribution
sns.histplot(df['Age'], kde=True)
plt.title('Age Distribution')
plt.show()

# Lung cancer cases by smoking history
sns.countplot(x='Smoking_History', hue='Lung_Cancer', data=df)
plt.title('Lung Cancer Cases by Smoking History')
plt.show()

# Correlation heatmap
corr = df.corr()
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
```
