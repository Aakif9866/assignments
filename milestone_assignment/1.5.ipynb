```python
tuple1 = (10, 20, "Apple", 3.4, 'a', ["master", "ji"], ("sita", "geeta", 22), [{"roll_no": 1}, {"name": "Navneet"}])

# a) Length of tuple1
print(len(tuple1))  

# b) Access "Navneet" from the tuple
print(tuple1[-1][1]["name"]) 

# c) Fetch value of roll_no from the tuple
print(tuple1[-1][0]["roll_no"])  

# d) Access the second element in the list within the tuple
print(tuple1[-3][1])  

# e) Fetch the element "22" from the tuple
print(tuple1[-2][-1]) 
```
