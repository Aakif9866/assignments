```python
class BankAccount:
    """
    Represents a base bank account with common attributes and methods.
    """

    def __init__(self, account_number, balance=0.0):
        self.__account_number = account_number
        self.__balance = balance

    def get_balance(self):
        """Returns the account balance."""
        return self.__balance

    def deposit(self, amount):
        """Deposits money into the account."""
        if amount > 0:
            self.__balance += amount
            print(f"Deposited {amount}. New balance: {self.get_balance()}")
        else:
            print("Invalid deposit amount.")

    def withdraw(self, amount):
        """Withdraws money from the account if sufficient funds are available."""
        if amount > 0 and self.__balance >= amount:
            self.__balance -= amount
            print(f"Withdrew {amount}. New balance: {self.get_balance()}")
        else:
            print("Insufficient funds for withdrawal.")


class SavingsAccount(BankAccount):
    """
    Represents a savings account with an interest rate.
    """

    def __init__(self, account_number, balance=0.0, interest_rate=0.03):
        super().__init__(account_number, balance)
        self.__interest_rate = interest_rate

    def calculate_interest(self):
        """Calculates interest on the account balance."""
        interest = self.get_balance() * self.__interest_rate
        self.__balance += interest
        print(f"Interest earned: {interest}. New balance: {self.get_balance()}")


class CheckingAccount(BankAccount):
    """
    Represents a checking account with a monthly fee.
    """

    def __init__(self, account_number, balance=0.0, monthly_fee=5.0):
        super().__init__(account_number, balance)
        self.__monthly_fee = monthly_fee

    def apply_monthly_fee(self):
        """Applies the monthly fee to the account."""
        if self.get_balance() >= self.__monthly_fee:
            self.__balance -= self.__monthly_fee
            print(f"Monthly fee applied: {self.__monthly_fee}. New balance: {self.get_balance()}")
        else:
            print("Insufficient funds to apply monthly fee.")


# Create account instances and test functionality
savings_account = SavingsAccount(1234, 1000.0)
checking_account = CheckingAccount(5678, 500.0)

savings_account.deposit(500.0)
savings_account.calculate_interest()

checking_account.withdraw(200.0)
checking_account.apply_monthly_fee()
```
