```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
df = pd.read_csv('presidential_election_polls_2024.csv.data')
```

```python
# Display the first few rows of the dataset
print(df.head())

# Check for missing values
print(df.isnull().sum())

# Summary statistics
print(df.describe())

# Check the data types
print(df.info())
```

```python
# Handling missing values (if any)
df.fillna(df.mean(), inplace=True)  # Example of filling missing values with mean
```

```python
# Preferred candidate by age group
sns.boxplot(x='Age_Group', y='Preferred_Candidate', data=df)
plt.title('Preferred Candidate by Age Group')
plt.show()

# Gender distribution of voters
sns.countplot(x='Gender', hue='Preferred_Candidate', data=df)
plt.title('Gender Distribution of Voters')
plt.show()

# Correlation heatmap
corr = df.corr()
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
```
