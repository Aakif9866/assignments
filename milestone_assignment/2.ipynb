```python
class Student:
    """
    Represents a student with encapsulated attributes and methods.
    """

    def __init__(self, name, age, roll_number):
        self.__name = name  # Encapsulated attribute for name
        self.__age = age    # Encapsulated attribute for age
        self.__roll_number = roll_number  # Encapsulated attribute for roll number

    def get_name(self):
        """Returns the student's name."""
        return self.__name

    def set_name(self, new_name):
        """Updates the student's name."""
        self.__name = new_name

    def get_age(self):
        """Returns the student's age."""
        return self.__age

    def set_age(self, new_age):
        """Updates the student's age."""
        self.__age = new_age

    def get_roll_number(self):
        """Returns the student's roll number."""
        return self.__roll_number

    def display_info(self):
        """Displays the student's information."""
        print(f"Name: {self.get_name()}")
        print(f"Age: {self.get_age()}")
        print(f"Roll Number: {self.get_roll_number()}")

    def update_details(self, new_name=None, new_age=None):
        """Updates the student's name and/or age."""
        if new_name is not None:
            self.set_name(new_name)
        if new_age is not None:
            self.set_age(new_age)

# Create student instances and test functionality
student1 = Student("Alice", 20, 123)
student1.display_info()  # Output: Name: Alice, Age: 20, Roll Number: 123

student1.update_details(new_age=21)
student1.display_info()  # Output: Name: Alice, Age: 21, Roll Number: 123
```
