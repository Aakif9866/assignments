```python
class Animal:
    """
    Represents a base animal with a method to make a sound.
    """

    def make_sound(self):
        """Makes a generic animal sound."""
        print("Generic animal sound")


class Dog(Animal):
    """
    Represents a dog that barks.
    """

    def make_sound(self):
        """Makes a dog bark."""
        print("Woof!")


class Cat(Animal):
    """
    Represents a cat that meows.
    """

    def make_sound(self):
        """Makes a cat meow."""
        print("Meow!")


# Create animal instances and test functionality
dog = Dog()
cat = Cat()

dog.make_sound()
cat.make_sound()
```
