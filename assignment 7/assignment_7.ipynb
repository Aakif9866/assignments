```python
import random
import statistics
import numpy as np
from scipy.stats import trim_mean

# Generate random integers between 90 and 130 (correct the range or sample size)
int_list = random.sample(range(90, 131), 41)  # Adjust sample size to match the range

# (i) Mean
def calculate_mean(numbers):
    return sum(numbers) / len(numbers)

# Median
def calculate_median(numbers):
    return statistics.median(numbers)

# Mode
def calculate_mode(numbers):
    return statistics.mode(numbers)

# Weighted Mean
def weighted_mean(values, weights):
    return sum(v * w for v, w in zip(values, weights)) / sum(weights)

# Geometric Mean
def geometric_mean(numbers):
    product = np.prod(numbers)
    return product ** (1 / len(numbers))

# Harmonic Mean
def harmonic_mean(numbers):
    return len(numbers) / sum(1 / x for x in numbers)

# Midrange
def midrange(numbers):
    return (max(numbers) + min(numbers)) / 2

# Trimmed Mean (exclude top 5% and bottom 5%)
def trimmed_mean(numbers, percent):
    return trim_mean(numbers, percent)

# Calculating all values
mean = calculate_mean(int_list)
median = calculate_median(int_list)
mode = calculate_mode(int_list)
weights = np.random.random(size=41)  # Adjust weight size to match the sample size
weighted_mean_value = weighted_mean(int_list, weights)
geo_mean = geometric_mean(int_list)
harmonic_mean_value = harmonic_mean(int_list)
midrange_value = midrange(int_list)
trim_mean_value = trimmed_mean(int_list, 0.05)

# Print the results
print(f"Mean: {mean}, Median: {median}, Mode: {mode}")
print(f"Weighted Mean: {weighted_mean_value}")
print(f"Geometric Mean: {geo_mean}, Harmonic Mean: {harmonic_mean_value}")
print(f"Midrange: {midrange_value}, Trimmed Mean: {trim_mean_value}")

```

```python
# Q2 Generate a list of 500 random integers between 200 and 300 and visualize the data:


import random
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
import statistics

# Generate list of random integers with replacement
int_list2 = random.choices(range(200, 301), k=500)

# (i) Visualizations
# 1. Frequency & Gaussian Distribution
sns.histplot(int_list2, kde=False, color="blue")
sns.kdeplot(np.random.normal(np.mean(int_list2), np.std(int_list2), len(int_list2)), color="red")
plt.show()

# 2. Frequency smoothed KDE plot
sns.kdeplot(int_list2, color="green")
plt.show()

# (ii) Range
def calculate_range(numbers):
    return max(numbers) - min(numbers)

# Variance and Standard Deviation
def calculate_variance(numbers):
    return statistics.variance(numbers)

def calculate_std_dev(numbers):
    return statistics.stdev(numbers)

# Interquartile Range (IQR)
def calculate_iqr(numbers):
    q75, q25 = np.percentile(numbers, [75, 25])
    return q75 - q25

# Coefficient of Variation
def coefficient_of_variation(numbers):
    return calculate_std_dev(numbers) / calculate_mean(numbers)

# Mean Absolute Deviation (MAD)
def mean_absolute_deviation(numbers):
    mean_val = calculate_mean(numbers)
    return sum(abs(x - mean_val) for x in numbers) / len(numbers)

# Quartile Deviation
def quartile_deviation(numbers):
    q1, q3 = np.percentile(numbers, [25, 75])
    return (q3 - q1) / 2

# Range-based Coefficient of Dispersion
def range_based_coefficient_of_dispersion(numbers):
    return calculate_range(numbers) / (max(numbers) + min(numbers))

# (iii) Calculation of required statistics
range_value = calculate_range(int_list2)
variance_value = calculate_variance(int_list2)
std_dev_value = calculate_std_dev(int_list2)
iqr_value = calculate_iqr(int_list2)
cv_value = coefficient_of_variation(int_list2)
mad_value = mean_absolute_deviation(int_list2)
qd_value = quartile_deviation(int_list2)
rcd_value = range_based_coefficient_of_dispersion(int_list2)

print(f"Range: {range_value}, Variance: {variance_value}, Std Dev: {std_dev_value}")
print(f"IQR: {iqr_value}, CV: {cv_value}, MAD: {mad_value}, QD: {qd_value}, RCD: {rcd_value}")
```

```python
# 3 Discrete Random Variable Class:

class DiscreteRandomVariable:
    def __init__(self, outcomes, probabilities):
        self.outcomes = outcomes
        self.probabilities = probabilities

    def expected_value(self):
        return sum(o * p for o, p in zip(self.outcomes, self.probabilities))

    def variance(self):
        mean = self.expected_value()
        return sum(p * (o - mean) ** 2 for o, p in zip(self.outcomes, self.probabilities))

# Example usage:
outcomes = [1, 2, 3, 4, 5]
probabilities = [0.1, 0.2, 0.3, 0.2, 0.2]
drv = DiscreteRandomVariable(outcomes, probabilities)
expected_val = drv.expected_value()
variance_val = drv.variance()

print(f"Expected Value: {expected_val}, Variance: {variance_val}")
```

```python
# 4 Simulate rolling a 6-sided die:

def roll_die(n_rolls=1000):
    outcomes = np.random.randint(1, 7, size=n_rolls)
    expected_value = np.mean(outcomes)
    variance = np.var(outcomes)
    return expected_value, variance

ev, var = roll_die()
print(f"Expected Value: {ev}, Variance: {var}")
```

```python
# 5 Generate random samples from a binomial distribution:
def generate_binomial_samples(n_trials, p_success, n_samples=1000):
    samples = np.random.binomial(n_trials, p_success, size=n_samples)
    mean = np.mean(samples)
    variance = np.var(samples)
    return mean, variance

mean_binom, var_binom = generate_binomial_samples(10, 0.5)
print(f"Binomial Distribution Mean: {mean_binom}, Variance: {var_binom}")
```

```python
# 6 Generate random numbers from a Gaussian distribution:
def generate_gaussian_samples(mean, std_dev, n_samples=1000):
    samples = np.random.normal(mean, std_dev, size=n_samples)
    mean_val = np.mean(samples)
    variance_val = np.var(samples)
    std_dev_val = np.std(samples)
    return mean_val, variance_val, std_dev_val

mean_gauss, var_gauss, std_gauss = generate_gaussian_samples(0, 1)
print(f"Gaussian Mean: {mean_gauss}, Variance: {var_gauss}, Std Dev: {std_gauss}")
```

```python
# 9 Z-Test
from scipy import stats

# Example data
sample_data = np.random.normal(5, 1, 50)
population_mean = 6

# Perform one-sample Z-Test
z_stat, p_value_ztest = stats.ttest_1samp(sample_data, population_mean)
print("Z-statistic:", z_stat)
print("P-value:", p_value_ztest)
```

```python
# 10 One-sample T-Test
# Example data for T-test
t_test_sample = np.random.normal(50, 10, 30)
pop_mean_ttest = 60

# Perform one-sample T-Test
t_stat, p_value_ttest = stats.ttest_1samp(t_test_sample, pop_mean_ttest)

print("T-statistic:", t_stat)
print("P-value:", p_value_ttest)
```

```python
# 11
# Two sample data
sample1 = np.random.normal(5, 1, 30)
sample2 = np.random.normal(6, 1, 30)

# Perform two-sample T-Test
t_stat_two, p_value_two = stats.ttest_ind(sample1, sample2)

print("T-statistic:", t_stat_two)
print("P-value:", p_value_two)

```

```python
# 12
# Paired data
before = np.random.normal(70, 5, 30)
after = np.random.normal(75, 5, 30)

# Perform paired T-Test
t_stat_paired, p_value_paired = stats.ttest_rel(before, after)

print("T-statistic:", t_stat_paired)
print("P-value:", p_value_paired)
```

```python
# 13
# Observed data for Chi-Square Test
observed = np.array([[10, 20, 30], [6, 9, 17]])

# Perform Chi-Square Test
chi2_stat, p_value_chi, dof, expected = stats.chi2_contingency(observed)

print("Chi2-statistic:", chi2_stat)
print("P-value:", p_value_chi)
```

```python
# 14
# Expected frequencies (based on Poisson distribution)
observed = [10, 12, 15, 7]
expected = [9, 13, 14, 8]

# Perform Chi-Square Goodness of Fit Test
chi2_stat_fit, p_value_fit = stats.chisquare(observed, expected)

print("Chi2-statistic:", chi2_stat_fit)
print("P-value:", p_value_fit)
```
